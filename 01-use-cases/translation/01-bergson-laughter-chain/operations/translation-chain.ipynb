# Translation as a chain of decisions

**Bergson, *Le Rire* · French → English · AI Humanities Lab**

This notebook adapts `translation_chained_prompts.ipynb` into a reproducible
workshop example. It shares the plain-text prompts and Python runner in this
folder. The French source stays present during every revision; the published
English comparison enters only at stage 7.

**Default: inspect the saved live example.** No API key or network call is
needed. Set `MODE = "live"` only when you want a fresh, billed API run. Put the
key in the repository's gitignored `.env`, never in a notebook cell.

Read [operations/README.md](README.md) for installation. Run this notebook from
inside the downloaded repository; a standalone copy has no inputs or prompts.


````python
from pathlib import Path
import sys
from IPython.display import display, Markdown, HTML

# Works from the repository root, this notebook's folder, or Jupyter's subfolders.
candidates = []
for base in [Path.cwd(), *Path.cwd().parents]:
    candidates.extend([base, base / "01-use-cases" / "translation"])
CASE = next((p for p in candidates if (p / "inputs/source-fr.md").exists()), None)
if CASE is None:
    raise FileNotFoundError("Open this notebook from inside the workshop repository.")
sys.path.insert(0, str(CASE / "operations"))
from chain import read_inputs, run_chain

MODE = "example"              # "live" makes new API calls
MODEL = "claude-sonnet-5-5"    # or another model enabled for your account
ROUNDS = 1                    # each extra round adds critique + revision
inputs = read_inputs()
````

## Inputs: establish what is actually being translated

The source is an excerpt from a 1938 French edition. The comparison is a 1911
English translation. Six editorial IDs keep the alignment visible. The
editorial brief is an input you can change; doing so requires a fresh run.


````python
display(Markdown(inputs["source"]))
display(Markdown(inputs["brief"]))
````

## Run once, then inspect the chain

Example mode validates and opens the saved run. Live mode executes the nine
stages and writes a new folder under `outputs/runs/`; it does not overwrite
the example. Each later request receives prior results. The two-step feedback
loop can be repeated with `ROUNDS` (1–3). Inspect the runner to change dependencies.


````python
run_dir = run_chain(mode=MODE, model=MODEL, rounds=ROUNDS)
print("Run folder:", run_dir.relative_to(CASE))

def show_stage(stage, filename=None):
    display(Markdown((CASE / "operations/prompts" / (stage + ".md")).read_text()))
    artifact = run_dir / (filename or stage + ".md")
    text = artifact.read_text()
    if artifact.suffix == ".json":
        text = "```json\n" + text + "\n```"
    display(Markdown("### Saved result\n\n" + text))
````

## 1 · Source check

A documented source record is different from a fresh web search.

````python
show_stage('01-source-check', None)
````

## 2 · Working lexicon

A short list of consequential choices is more useful here than speculative etymologies for every token.

````python
show_stage('02-lexicon', None)
````

## 3 · Context and rhythm

How do the four questions, the evasive verbs, and the living-growth image affect an English rendering?

````python
show_stage('03-context-rhythm', None)
````

## 4 · First draft

The published English translation has not entered the prompt yet.

````python
show_stage('04-draft', None)
````

## 5–6 · Critique and revision

The model critic reads the French too. This is self-critique, not independent verification.

````python
show_stage('05-critique', '05-critique-1.md')
````

````python
show_stage("06-revise", "06-revise-1.md")
# Additional rounds, when present, have numbered files too.
for round_number in range(2, ROUNDS + 1):
    show_stage("05-critique", f"05-critique-{round_number}.md")
    show_stage("06-revise", f"06-revise-{round_number}.md")
````

## 7 · Published comparison

Now the actual Brereton/Rothwell excerpt enters. A title and URL would not be enough to compare wording.

````python
show_stage('07-compare', None)
````

## 8 · Editorial decisions

The final pass still receives the French. Six segment IDs and a short-excerpt word budget keep the output readable.

````python
show_stage('08-final-edit', '08-final-edit.json')
````

## 9 · Annotation

Four notes quote exact phrases in both languages. Structural checks do not establish that the interpretation is correct.

````python
show_stage('09-annotate', '09-annotate.json')
````
